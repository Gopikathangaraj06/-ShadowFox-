```python
# Student mark analysis
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Name": ["Asha", "Rahul", "Kiran", "Meena", "Arjun"],
    "Math": [78, 85, 90, 88, 76],
    "Science": [82, 89, 91, 84, 80],
    "English": [75, 80, 85, 78, 82]
}

df = pd.DataFrame(data)
df
```

```python
df.info()
```

```python
average_scores = df[["Math", "Science", "English"]].mean()
average_scores
```

```python
average_scores.plot(kind="bar")
plt.title("Average Scores by Subject")
plt.ylabel("Average Marks")
plt.xlabel("Subject")
plt.show()
```

```python
## Conclusion
#From the analysis, we observe that **Science** has the highest avg

```
